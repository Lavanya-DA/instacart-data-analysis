```python
prior = pd.read_csv(
    "order_products__prior.csv",
    usecols=['order_id', 'product_id', 'reordered']
)
```

```python
customer_frequency = orders.groupby('user_id')['order_number'].max()

customer_segments = pd.cut(
    customer_frequency,
    bins=[3, 6, 20, 100],
    labels=['Low Frequency', 'Medium Frequency', 'High Frequency']
)

orders['segment'] = orders['user_id'].map(customer_segments)
```

```python
product_reorder = (prior.groupby('product_id')['reordered'].mean().sort_values(ascending=False))
print(product_reorder.head(10))
```

```python
product_stats = (
    prior.groupby('product_id')
    .agg(
        purchases=('reordered', 'count'),
        reorder_rate=('reordered', 'mean')
    )
)

product_stats = product_stats[
    product_stats['purchases'] >= 1000
]

product_stats['reorder_rate'] = (
    product_stats['reorder_rate'] * 100
).round(2)

print(
    product_stats
    .sort_values('reorder_rate', ascending=False)
    .head(10)
)
```

```python
top_reorder_products = (
    product_stats
    .sort_values('reorder_rate', ascending=False)
    .head(10)
    .reset_index()
)

top_reorder_products = top_reorder_products.merge(
    products[['product_id', 'product_name']],
    on='product_id',
    how='left'
)

print(
    top_reorder_products[
        ['product_id', 'product_name', 'purchases', 'reorder_rate']
    ]
)
```

```python
prior_cart = pd.read_csv(
    "order_products__prior.csv",
    usecols=['order_id', 'add_to_cart_order']
)
```
